```python
#selecting 5-6 features that represent customer demographic
#Age - Income - Total_Spending - NumWebPurchases - NumStorePurchases - NumWebVisitsMonth - Recency
```

```python
features = ["Age", "Income", "Total_Spending", "NumWebPurchases", "NumStorePurchases","NumWebVisitsMonth","Recency"]
```

```python
X = df[features].copy()
```

```python
X
```

```python
#preprocessing with scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
#using k means for the clustering algorithm
from sklearn.cluster import KMeans
```

```python
wcss = []
```

```python
for i in range (2,10):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
#imported again because it was causing error with str
import importlib
import matplotlib.pyplot as plt

importlib.reload(plt)
```

```python
#plotting this
plt.plot(range(2,10),wcss,marker = "o")
plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show
```

```python
kmeans = KMeans(n_clusters = 6)
df["Cluster"] = kmeans.fit_predict(X_scaled)
```

```python
df["Cluster"]
```

```python
df.head()
```

```python
#analysing the cluster
cluster_summary = df.groupby("Cluster")[features].mean()
```

```python
cluster_summary
```

```python
df["Cluster"].value_counts()
```

```python
#visualizing the cluster principal component analysis
from sklearn.decomposition import PCA

pca = PCA(n_components= 2)
pca_data = pca.fit_transform(X_scaled)
df["PCA1"], df["PCA2"] = pca_data[:,0], pca_data[:,1]
```

```python
pca_data
```

```python
df["PCA1"]
```

```python
#creating a scatter plot
sns.scatterplot(x="PCA1", y="PCA2", hue = "Cluster", data = df, palette = "Set1")
plt.title("Customer Segmentation (PCA)")
plt.show()
```
